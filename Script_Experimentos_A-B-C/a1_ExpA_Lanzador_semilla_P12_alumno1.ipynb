```python
from pathlib import Path
import papermill as pm

# Carpeta donde están los notebooks originales
carpeta = Path("/home/ds/buckets/b1/labo2026v/ExpColaborativos")

# Los tres workflows originales
workflows = [
    "a1_V01_P12_ExpA_620_WorkFlow_01_junior_grupoA.ipynb"
]
```

```python
for workflow in workflows:
    print(workflow, (carpeta / workflow).exists())
```

Comprobacion de etiquetado *parameters*

```python
import json

archivos = [
    "a1_V01_P12_ExpA_620_WorkFlow_01_junior_grupoA.ipynb",
    "a1_V01_P12_ExpB_620_WorkFlow_01_junior_grupoA.ipynb",
    "a1_V01_P12_ExpC_620_WorkFlow_01_junior_grupoA.ipynb"
]

for archivo in archivos:
    with open(carpeta / archivo, encoding="utf-8") as f:
        notebook = json.load(f)

    etiquetas = [
        celda for celda in notebook["cells"]
        if "parameters" in celda.get("metadata", {}).get("tags", [])
    ]

    print(archivo, "→", "OK" if len(etiquetas) == 1 else "REVISAR")
```

aca recien se puede poner a correr el resto

```python
# Cinco semillas del alumno 1
semillas = [
    193357, 826351, 143249, 648931, 268693
]

# Carpeta independiente para los notebooks ejecutados
salida = carpeta / "ejecutados_alumno1"
salida.mkdir(parents=True, exist_ok=True)

# Primer número de experimento
experimento_base = 9300    #se cambia a 9300 para que no haya problemas de sobre escritura entre alumno

import time

inicio = time.time()

# Ejecutar el workflow para cada semilla
for i, semilla in enumerate(semillas):

    for j, workflow in enumerate(workflows):

        experimento = experimento_base + i * len(workflows) + j

        notebook_entrada = carpeta / workflow

        notebook_salida = salida / (
            f"{Path(workflow).stem}"
            f"_semilla_{semilla}"
            f"_exp_{experimento}.ipynb"
        )

        print(
            f"\nEjecutando {workflow} | "
            f"Semilla: {semilla} | "
            f"Experimento: {experimento}"
        )

        pm.execute_notebook(
            str(notebook_entrada),
            str(notebook_salida),
            parameters={
                "semilla_lanzador": semilla,
                "experimento_lanzador": experimento
            },
            progress_bar=False,
            kernel_name=None
        )

print("\nFinalizaron los 5 experimentos.")
```

```python
fin = time.time()

tiempo_total = (fin - inicio) / 3600

print(f"Tiempo total de ejecución: {tiempo_total:.2f} horas")
```
